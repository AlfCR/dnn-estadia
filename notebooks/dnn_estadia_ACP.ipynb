```python
# Celda 5: huellas SHA-256 de originales y respaldo
def calcular_sha256(ruta):
    h = hashlib.sha256()
    with open(ruta, 'rb') as f:
        # 1024 * 1024 para leer en bloques de 1 MB
        for bloque in iter(lambda: f.read(1024 * 1024), b''):
            h.update(bloque)
    return h.hexdigest()

filas = []
for carpeta in (carpeta1, carpeta2):
    nombre_carpeta = os.path.basename(carpeta)
    for nombre in sorted(os.listdir(carpeta)):
        ruta_original = os.path.join(carpeta, nombre)
        ruta_respaldo = os.path.join(respaldo_dataset, nombre_carpeta, nombre)
        filas.append({
            'carpeta': nombre_carpeta,
            'archivo': nombre,
            'tamano_kb': round(os.path.getsize(ruta_original) / 1024, 1),
            'sha256_original': calcular_sha256(ruta_original),
            'sha256_respaldo': calcular_sha256(ruta_respaldo),
        })

huellas = pd.DataFrame(filas)

# El respaldo debe ser idéntico al original
assert (huellas['sha256_original'] == huellas['sha256_respaldo']).all(), \
    "Hay archivos distintos entre el original y el respaldo"
print(f"Respaldo idéntico al original: {len(huellas)} archivos")

# Línea base: se guarda solo la primera vez; después, se compara contra ella
destino_huellas = os.path.join(ruta_salida, 'huellas_originales.csv')
if os.path.exists(destino_huellas):
    base_previa = pd.read_csv(destino_huellas)
    comp = huellas.merge(
        base_previa[['carpeta', 'archivo', 'sha256_original']],
        on=['carpeta', 'archivo'], how='outer', suffixes=('_hoy', '_base'))
    cambios = comp[comp['sha256_original_hoy'] != comp['sha256_original_base']]
    if cambios.empty:
        print("Los originales siguen idénticos a la línea base.")
    else:
        print(f"ATENCIÓN: archivos distintos o faltantes:\n{cambios}")
else:
    huellas.to_csv(destino_huellas, index=False)
    print(f"Línea base guardada en: {destino_huellas}")

# Vista corta (huella recortada a 12 caracteres)
huellas[['carpeta', 'archivo', 'tamano_kb']].assign(sha256=huellas['sha256_original'].str[:12])
```

```python
# Celda 6: Auditar contenido y formato (por bloques de encabezado)
import csv
import io
import os
from pathlib import Path
import pandas as pd

FALTANTES = {'', '-','nan', 'NaN', 'null', 'None'}

def normalizar_texto(valor):
    if pd.isna(valor):
        return ''
    return str(valor).replace('\u00a0', ' ').strip()

def leer_filas(ruta):

    #Se devuelve las filas como listas de texto, sin que pandas decida el ancho.
    if ruta.suffix.lower() == '.xlsx':
        bruto = pd.read_excel(ruta, header=None, dtype='string', keep_default_na=False)
        return bruto.fillna('').values.tolist(), 'xlsx: primera hoja'
    for cod in ('utf-8-sig', 'cp1252'):
        try:
            with open(ruta, 'r', encoding=cod, newline='') as f:
                texto = f.read()
            break
        except UnicodeDecodeError:
            continue
    else:
        raise ValueError('No se pudo determinar la codificación del CSV.')
    try:
        sep = csv.Sniffer().sniff(texto[:8192], delimiters=',;\t|').delimiter
    except csv.Error:
        sep = ','
    filas = list(csv.reader(io.StringIO(texto, newline=''), delimiter=sep))
    return filas, f'csv: separador={sep!r}, encoding={cod}'

def partir_en_bloques(filas):

     #Un bloque = un encabezado (el primero o uno repetido, como el caso de los datos de chiapas) y las filas que le siguen.
    pares = [(n, f) for n, f in enumerate(filas, start=1)
             if any(normalizar_texto(x) for x in f)]          # sin líneas en blanco
    if not pares:
        return []
    primer_campo = normalizar_texto(pares[0][1][0])
    bloques = []
    for n, fila in pares:
        if normalizar_texto(fila[0]) == primer_campo:
            bloques.append({'linea': n,
                            'encabezado': [normalizar_texto(x) for x in fila],
                            'filas': []})
        else:
            bloques[-1]['filas'].append(fila)
    return bloques

def auditar_bloques(bloques):
    vacia_por_columna = {}
    for b in bloques:
        for i, nombre in enumerate(b['encabezado']):
            etiqueta = nombre or f'__SIN_NOMBRE_{i + 1}'
            vacia = all((normalizar_texto(r[i]) in FALTANTES) if i < len(r) else True
                        for r in b['filas'])
            vacia_por_columna.setdefault(etiqueta, []).append(vacia)
    columnas = {n for n in vacia_por_columna if not n.startswith('__SIN_NOMBRE_')}
    vacias = sorted(n for n, v in vacia_por_columna.items() if all(v))
    duplicadas = sorted({n for b in bloques for n in b['encabezado']
                         if n and b['encabezado'].count(n) > 1})
    return columnas, vacias, duplicadas

esquemas, filas_reporte = {}, []
for carpeta in (carpeta1, carpeta2):
    for ruta in sorted(Path(carpeta).iterdir()):
        if (not ruta.is_file() or ruta.name.startswith('~$')
                or ruta.suffix.lower() not in ('.csv', '.xlsx')):
            continue
        clave = f'{ruta.parent.name}/{ruta.name}'
        try:
            filas, detalle = leer_filas(ruta)
            bloques = partir_en_bloques(filas)
            columnas, vacias, duplicadas = auditar_bloques(bloques)
            esquemas[clave] = columnas
            filas_reporte.append({
                'archivo': clave,
                'filas_datos': sum(len(b['filas']) for b in bloques),
                'n_bloques': len(bloques),
                'lineas_encabezado_repetido': [b['linea'] for b in bloques[1:]],
                'n_esquemas_distintos': len({tuple(b['encabezado']) for b in bloques}),
                'n_columnas': len(columnas),
                'columnas_duplicadas': duplicadas,
                'columnas_vacias': vacias,
                'lectura': detalle,
                'error': '',
            })
        except Exception as error:
            filas_reporte.append({'archivo': clave, 'error': str(error)})

# Columnas que le faltan a cada archivo respecto a la unión de SU carpeta (solo el conteo)
union_carpeta = {}
for clave, cols in esquemas.items():
    union_carpeta.setdefault(clave.split('/')[0], set()).update(cols)
for fila in filas_reporte:
    if fila['archivo'] in esquemas:
        nombre_carpeta = fila['archivo'].split('/')[0]
        fila['n_faltantes_vs_carpeta'] = len(union_carpeta[nombre_carpeta] - esquemas[fila['archivo']])

auditoria = pd.DataFrame(filas_reporte)
display(auditoria.drop(columns=['lectura']))

ruta_auditoria = os.path.join(ruta_salida, 'auditoria_archivos.csv')
auditoria.to_csv(ruta_auditoria, index=False)
print(f'Auditoría guardada en: {ruta_auditoria}')
```

```python
# Celda 7: detalle por bloque
import hashlib
import os
from pathlib import Path
import pandas as pd

# Para evitar comparar archivo por archivo, se calcula una hash binario
# que permite identificar las secciones concatenadas o repetidas entres diferentes archivos

def huella_bloque(b):
    h = hashlib.sha256()
    for r in b['filas']:
        h.update('|'.join(normalizar_texto(x) for x in r).encode('utf-8'))
        h.update(b'\n')
    return h.hexdigest()[:12]

registros = []
for carpeta in (carpeta1, carpeta2):
    for ruta in sorted(Path(carpeta).iterdir()):
        if ruta.suffix.lower() != '.csv':
            continue
        filas, _ = leer_filas(ruta)
        for k, b in enumerate(partir_en_bloques(filas), start=1):

            col_tiempo = next((idx for idx, c in enumerate(b['encabezado'])
                               if 'device time' in c.lower() or 'time' in c.lower()), None)

            if col_tiempo is not None:
                # Verificación de longitud segura para evitar IndexError
                tiempos = [normalizar_texto(f[col_tiempo]) for f in b['filas']
                           if col_tiempo < len(f) and normalizar_texto(f[col_tiempo])]
            else:
                tiempos = []

            registros.append({
                'archivo': ruta.name,
                'carpeta': ruta.parent.name,
                'bloque': k,
                'linea_encabezado': b['linea'],
                'n_columnas': len(b['encabezado']),
                'n_filas': len(b['filas']),
                'primer_tiempo': tiempos[0] if tiempos else '',
                'ultimo_tiempo': tiempos[-1] if tiempos else '',
                'huella_bloque': huella_bloque(b),
            })
bloques_df = pd.DataFrame(registros)
bloques_df['veces_repetido'] = bloques_df.groupby('huella_bloque')['huella_bloque'].transform('size')

unicos = bloques_df.drop_duplicates('huella_bloque')
print(f"Bloques: {len(bloques_df)} | únicos: {len(unicos)}")
print(f"Filas en todos los bloques: {bloques_df['n_filas'].sum()} | filas únicas: {unicos['n_filas'].sum()}")

display(bloques_df[bloques_df['carpeta'].str.contains('Chiapas')].drop(columns='carpeta'))
bloques_df.to_csv(os.path.join(ruta_salida, 'bloques_archivos.csv'), index=False)
```

```python
#Celda 8 catálogo de viajes únicos (un bloque por huella)
import hashlib
import json
from pathlib import Path

import pandas as pd


def huella_bloque(encabezado, filas):
    #SHA-256 del encabezado y los datos normalizados (filas rellenadas al ancho del encabezado).
    ancho = len(encabezado)
    filas_norm = []
    for fila in filas:
        if len(fila) > ancho:
            raise ValueError(f"Fila con {len(fila)} campos y encabezado con {ancho} columnas")
        filas_norm.append(fila + [''] * (ancho - len(fila)))
    contenido = json.dumps([encabezado, filas_norm], ensure_ascii=False, separators=(',', ':'))
    return hashlib.sha256(contenido.encode('utf-8')).hexdigest(), filas_norm


bloques_por_huella = {}
bloques_sin_filas = 0

for carpeta in (carpeta1, carpeta2):
    for ruta in sorted(Path(carpeta).iterdir()):
        # Solo CSV: el .xlsx es copia y puede alterar formatos de números y fechas
        if not ruta.is_file() or ruta.suffix.lower() != '.csv':
            continue

        filas_archivo, _ = leer_filas(ruta)

        for numero_bloque, bloque in enumerate(partir_en_bloques(filas_archivo), start=1):
            if not bloque['filas']:
                bloques_sin_filas += 1
                continue

            encabezado = [normalizar_texto(c) for c in bloque['encabezado']]
            filas = [[normalizar_texto(v) for v in fila] for fila in bloque['filas']]
            huella, filas = huella_bloque(encabezado, filas)

            if huella not in bloques_por_huella:
                i_t = encabezado.index('Device Time') if 'Device Time' in encabezado else None
                tiempos = [f[i_t] for f in filas if f[i_t]] if i_t is not None else []
                bloques_por_huella[huella] = {
                    'carpeta': ruta.parent.name,
                    'encabezado': encabezado,
                    'filas': filas,
                    'primera_fecha': tiempos[0] if tiempos else '',
                    'ultima_fecha': tiempos[-1] if tiempos else '',
                    'fuentes': [],
                }
            bloques_por_huella[huella]['fuentes'].append(
                f'{ruta.name} (bloque {numero_bloque})')

# CONCEPTO DE VIAJE:
# Un "viaje" representa un recorrido o sesión continua.
# En esta celda se desduplican los bloques de datos, asignando un identificador único (id_viaje) a cada uno de los 18 recorridos identificados.

# IDs: viaje_num consecutivo (orden de aparición) + id_viaje estable derivado de la huella
ids = {h: f'V{h[:16]}' for h in bloques_por_huella}
if len(set(ids.values())) != len(ids):
    raise ValueError('Colisión en el prefijo del id_viaje; usa el SHA-256 completo.')

catalogo_viajes = pd.DataFrame([
    {
        'viaje_num': n,
        'id_viaje': ids[h],
        'carpeta': b['carpeta'],
        'n_filas': len(b['filas']),
        'n_columnas': len(b['encabezado']),
        'primera_fecha': b['primera_fecha'],
        'ultima_fecha': b['ultima_fecha'],
        'huella_bloque': h,
        'fuentes': ' | '.join(sorted(set(b['fuentes']))),
    }
    for n, (h, b) in enumerate(bloques_por_huella.items(), start=1)
])

# Datos: solo las filas de cada viaje único, con id_viaje y carpeta (el resto va en el catálogo)
partes = []
for h, b in bloques_por_huella.items():
    if 'id_viaje' in b['encabezado'] or 'carpeta' in b['encabezado']:
        raise ValueError("El archivo ya trae una columna 'id_viaje' o 'carpeta'; renómbrala primero.")
    df = pd.DataFrame(b['filas'], columns=b['encabezado'])
    df.insert(0, 'carpeta', b['carpeta'])
    df.insert(0, 'id_viaje', ids[h])
    partes.append(df)

dataset_viajes_unicos = pd.concat(partes, ignore_index=True, sort=False)

Path(ruta_salida).mkdir(parents=True, exist_ok=True)
ruta_catalogo = Path(ruta_salida) / 'viajes_unicos.csv'
ruta_dataset = Path(ruta_salida) / 'datos_viajes_unicos.csv'
catalogo_viajes.to_csv(ruta_catalogo, index=False)
dataset_viajes_unicos.to_csv(ruta_dataset, index=False)

print(f"Bloques únicos: {len(catalogo_viajes)}  (esperado: 18)")
print(f"Filas conservadas: {len(dataset_viajes_unicos)}  (esperado: 30424)")
print(f"Bloques sin filas omitidos: {bloques_sin_filas}  (esperado: 0)")
print(f"Catálogo: {ruta_catalogo}")
print(f"Datos: {ruta_dataset}")
display(catalogo_viajes.drop(columns=['huella_bloque', 'fuentes']))
```

```python
#celda 9 - Inventario de columnas
print("Inventario de columnas")
from pathlib import Path
import pandas as pd

ruta_entrada = Path(ruta_salida) / "datos_viajes_unicos.csv"
df = pd.read_csv(ruta_entrada, dtype=str, keep_default_na=False)

if "id_viaje" not in df.columns:
    raise ValueError("No se encontró la columna id_viaje")

faltantes = {"", "-", "nan", "NaN"}
columnas_id = {"id_viaje", "carpeta"}
columnas_variables = [c for c in df.columns if c not in columnas_id]

# True = dato útil (el cero cuenta como dato)
utiles = df[columnas_variables].apply(lambda col: ~col.isin(faltantes))

# % de filas útiles por viaje y columna
pct_viaje = utiles.groupby(df["id_viaje"]).mean() * 100

# Carpeta de cada viaje
carpeta_viaje = df.drop_duplicates("id_viaje").set_index("id_viaje")["carpeta"]
es_chiapas = carpeta_viaje.reindex(pct_viaje.index).str.contains("Chiapas")

# Valores distintos entre las filas útiles (para detectar constantes)
n_distintos = {c: df.loc[utiles[c], c].nunique() for c in columnas_variables}

inventario = pd.DataFrame({
    "viajes_con_dato": (pct_viaje > 0).sum(),
    "viajes_measurement": (pct_viaje[~es_chiapas] > 0).sum(),
    "viajes_chiapas": (pct_viaje[es_chiapas] > 0).sum(),
    "pct_filas_util_global": (utiles.mean() * 100).round(1),
    "valores_distintos": pd.Series(n_distintos),
})
inventario["constante"] = (inventario["valores_distintos"] == 1)
inventario["sin_datos"] = (inventario["viajes_con_dato"] == 0)
inventario = (inventario.sort_values(["viajes_con_dato", "pct_filas_util_global"],
                                     ascending=False)
              .rename_axis("columna").reset_index())

inventario.to_csv(Path(ruta_salida) / "inventario_columnas.csv", index=False)

print(f"Columnas: {len(inventario)} | sin datos: {inventario['sin_datos'].sum()} | "
      f"constantes: {inventario['constante'].sum()} | "
      f"con datos y variables: {(~inventario['sin_datos'] & ~inventario['constante']).sum()}")
display(inventario[~inventario["sin_datos"]].head(40))
```

```python
# Celda 10 -> Limpieza: de df (texto) a un dataset numérico, con unidades y nombres uniformes
MESES = {'ene': 1, 'feb': 2, 'mar': 3, 'abr': 4, 'may': 5, 'jun': 6,
         'jul': 7, 'ago': 8, 'sep': 9, 'oct': 10, 'nov': 11, 'dic': 12}

def convertir_fecha(serie):
    """'12-nov-2022 20:05:18.394' o '10-nov.-2022 09:24:07.144' -> datetime."""
    ex = serie.str.lower().str.extract(r'(\d{1,2})-([a-z]+)\.?-(\d{4})\s+(\d{2}:\d{2}:\d{2})(?:\.(\d+))?')
    mes = ex[1].map(MESES).astype('Int64').astype(str).str.zfill(2)
    texto = ex[2] + '-' + mes + '-' + ex[0].str.zfill(2) + ' ' + ex[3] + '.' + ex[4].fillna('0')
    return pd.to_datetime(texto, errors='coerce')

def fahrenheit_a_celsius(temp_f): return (temp_f - 32) * (5 / 9)
def mph_a_kmh(velocidad_mph):     return velocidad_mph * 1.609344
def psi_a_kpa(presion_psi):       return presion_psi * 6.894757
def ms_a_kmh(velocidad_ms):       return velocidad_ms * 3.6
def sin_cambio(x):                return x

conversiones = {
    'Engine RPM(rpm)':                  ('rpm', sin_cambio),
    'Engine Load(%)':                   ('carga_motor_pct', sin_cambio),
    'Engine Coolant Temperature(°F)':   ('temp_refrigerante_C', fahrenheit_a_celsius),
    'Intake Air Temperature(°F)':       ('temp_aire_admision_C', fahrenheit_a_celsius),
    'Intake Manifold Pressure(psi)':    ('presion_multiple_kPa', psi_a_kpa),
    'Throttle Position(Manifold)(%)':   ('acelerador_manifold_pct', sin_cambio),
    'Absolute Throttle Position B(%)':  ('acelerador_b_pct', sin_cambio),  # respaldo para el viaje 8
    'Air Fuel Ratio(Measured)(:1)':     ('afr', sin_cambio),
    'Speed (OBD)(mph)':                 ('velocidad_obd_kmh', mph_a_kmh),
    'Voltage (OBD Adapter)(V)':         ('voltaje_adaptador_V', sin_cambio),
    'GPS Speed (Meters/second)':        ('velocidad_gps_kmh', ms_a_kmh),
    'G(x)':                             ('g_x_ms2', sin_cambio),
    'G(y)':                             ('g_y_ms2', sin_cambio),
    'G(z)':                             ('g_z_ms2', sin_cambio),
}
# Variables del dataset (las convertidas + la aceleración total derivada)
variables_nuevas = [nuevo for nuevo, _ in conversiones.values()] + ['acel_total_ms2']

# df viene del catalogo_viajes
assert {'id_viaje', 'carpeta', 'Device Time'} <= set(df.columns), "df no es el de la celda 10"
assert df['id_viaje'].nunique() == 18, "df no tiene los 18 viajes únicos"
faltan = [c for c in conversiones if c not in df.columns]
assert not faltan, f"Faltan columnas en df: {faltan}"

limpio = df[['id_viaje', 'carpeta', 'Device Time'] + list(conversiones)].copy()
limpio['timestamp'] = convertir_fecha(limpio['Device Time'])
assert limpio['timestamp'].notna().all(), "Hay fechas que no se pudieron convertir"

dataset = pd.DataFrame({
    'viaje_num': limpio['id_viaje'].map(catalogo_viajes.set_index('id_viaje')['viaje_num']),
    'id_viaje': limpio['id_viaje'],
    'fuente': np.where(limpio['carpeta'].str.contains('Chiapas'), 'Chiapas', 'Complementary'),
    'timestamp': limpio['timestamp'],
})
for original, (nuevo, funcion) in conversiones.items():
    # '-'  pasan a NaN (valores vacíos)
    dataset[nuevo] = funcion(pd.to_numeric(limpio[original], errors='coerce'))

# Aceleración total sin gravedad: no depende de cómo estaba colocado el teléfono
dataset['acel_total_ms2'] = (np.sqrt(dataset['g_x_ms2']**2 + dataset['g_y_ms2']**2
                                     + dataset['g_z_ms2']**2) - 9.80665)

dataset[variables_nuevas] = dataset[variables_nuevas].round(2)
dataset['tiene_motor'] = dataset.groupby('viaje_num')['rpm'].transform(lambda s: s.notna().any())

antes = len(dataset)
dataset = dataset.drop_duplicates(subset=['viaje_num', 'timestamp'])
dataset = dataset.sort_values(['viaje_num', 'timestamp']).reset_index(drop=True)
print(f"Filas: {len(dataset)} (duplicados quitados: {antes - len(dataset)}) | viajes: {dataset['viaje_num'].nunique()}")

dataset.to_csv(os.path.join(ruta_salida, 'dataset_unificado.csv'), index=False)
display(dataset.head())
```

```python
#Celda 12
import pandas as pd
import numpy as np
import os

TIEMPO_RECORTE_SEG = 10    # supuesto: se descartan los primeros segundos de cada viaje
HUECO_MAX_SEG = 5          # un salto mayor parte el viaje en otro segmento
RANGO_INTERPOLACION = 3    # máximo de muestras vacías consecutivas que se rellenan
DURACION_MINIMA_SEG = 30   # segmentos más cortos se descartan

ruta_dataset = os.path.join(ruta_salida, 'dataset_unificado.csv')
df_raw = pd.read_csv(ruta_dataset, parse_dates=['timestamp'])

columnas_meta = ['viaje_num', 'id_viaje', 'fuente', 'tiene_motor', 'segmento_id']
columnas_num = [c for c in df_raw.columns if c not in columnas_meta + ['timestamp']]

# 1. Recorte inicial de cada viaje
inicio = df_raw.groupby('viaje_num')['timestamp'].transform('min')
df_recorte = df_raw[df_raw['timestamp'] >= inicio + pd.Timedelta(seconds=TIEMPO_RECORTE_SEG)].copy()
print(f"Filas: {len(df_raw)} -> {len(df_recorte)} (quitadas: {len(df_raw) - len(df_recorte)})")

# 2. Segmentos: un salto mayor a HUECO_MAX_SEG abre un segmento nuevo
df_recorte = df_recorte.sort_values(['viaje_num', 'timestamp']).reset_index(drop=True)
dt = df_recorte.groupby('viaje_num')['timestamp'].diff().dt.total_seconds()
salto = (dt > HUECO_MAX_SEG) | dt.isna()
df_recorte['segmento_id'] = salto.groupby(df_recorte['viaje_num']).cumsum()

# 3. Remuestreo a 1 Hz: promedio de lo que cae en cada segundo
df_recorte['timestamp_1s'] = df_recorte['timestamp'].dt.floor('1s')
df_1hz = (df_recorte.groupby(columnas_meta + ['timestamp_1s'])[columnas_num].mean()
          .reset_index().rename(columns={'timestamp_1s': 'timestamp'}))

# 4. Interpolación de vacíos cortos, segmento por segmento
partes = []
for (viaje, seg), g in df_1hz.groupby(['viaje_num', 'segmento_id']):
    meta = g.iloc[0][columnas_meta]           # los metadatos son constantes dentro del segmento
    g = g.set_index('timestamp').resample('1s').asfreq()
    for c in columnas_meta:
        g[c] = meta[c]
    g[columnas_num] = g[columnas_num].interpolate(method='linear', limit=RANGO_INTERPOLACION)
    partes.append(g.reset_index())
df_interpolado = pd.concat(partes, ignore_index=True)
df_interpolado['tiene_motor'] = df_interpolado['tiene_motor'].astype(bool)
df_interpolado[['viaje_num', 'segmento_id']] = df_interpolado[['viaje_num', 'segmento_id']].astype(int)

# 5. Descartar segmentos cortos (a 1 Hz, filas = segundos)
filas_seg = df_interpolado.groupby(['viaje_num', 'segmento_id'])['timestamp'].transform('size')
df_final = df_interpolado[filas_seg >= DURACION_MINIMA_SEG].copy()
df_final[columnas_num] = df_final[columnas_num].round(2)

# Guardar y reportar
df_final.to_csv(os.path.join(ruta_salida, 'dataset_1hz.csv'), index=False)

print(f"\nFilas finales: {len(df_final)} | viajes: {df_final['viaje_num'].nunique()} | "
      f"segmentos: {df_final.groupby(['viaje_num', 'segmento_id']).ngroups}")
resumen = df_final.groupby('viaje_num').agg(filas=('timestamp', 'size'),
                                            segmentos=('segmento_id', 'nunique'))
resumen['segmento_mas_largo_s'] = (df_final.groupby(['viaje_num', 'segmento_id']).size()
                                   .groupby('viaje_num').max())
display(resumen)
print("\nVacíos que quedaron en viajes con motor:")
print(df_final.loc[df_final['tiene_motor'], ['rpm', 'carga_motor_pct']].isna().sum())
```

```python
#Celda 13 -> División por viaje
import pandas as pd
import os

df = pd.read_csv(os.path.join(ruta_salida, 'dataset_1hz.csv'), parse_dates=['timestamp'])

#Definir diccionario
particiones_config = {
    'entrenamiento': [1, 2, 3, 4, 8, 9],
    'validacion':[7],
    'prueba':[5,6],
    'prueba_externa':[16],
    'sin_motor':[10, 11, 12, 13, 15, 18],

}

# Comprobación 1: ningún viaje aparece en dos particiones ()
todos = [v for viajes in particiones_config.values() for v in viajes]
assert len(todos) == len(set(todos)), "Hay viajes asignados a más de una partición."

#Invertir el diccionario
mapa_viajes = {v: p for p, viajes in particiones_config.items() for v in viajes}
df['particion'] = df['viaje_num'].map(mapa_viajes)

#comprobación 2: ninguna fila quedó sin partición
# En caso de que un viaje est sin asignar sale como NaN
assert df['particion'].notna().all(), (
     f"viajes sin asignar: {sorted(df.loc[df['particion'].isna(), 'viaje_num'].unique())}"
)

#Comprobación 3: los 9 viajes con motor están en entrenamiento, , validación o prueba
viajes_motor = {1, 2, 3, 4, 5, 6, 7, 8, 9}
viajes_modelo = set(particiones_config['entrenamiento'] + particiones_config['validacion'] + particiones_config['prueba'])
assert viajes_motor == viajes_modelo, "Los viajes del modelo no coinciden con los 9 viajes con motor."

# Resumen: viajes, filas y porcentaje (calculado sobre las filas del modelo)
resumen = df.groupby('particion').agg(
    viajes=('viaje_num', lambda x: sorted(x.unique())),
    n_viajes=('viaje_num', 'nunique'),
    filas=('viaje_num', 'size'),
).reindex(list(particiones_config))
filas_modelo = resumen.loc[['entrenamiento', 'validacion', 'prueba'], 'filas'].sum()
resumen['pct_filas_modelo'] = (100 * resumen['filas'] / filas_modelo).round(1)
resumen.loc[['prueba_externa', 'sin_motor'], 'pct_filas_modelo'] = float('nan')
display(resumen)

df.to_csv(os.path.join(ruta_salida, 'dataset_1hz_particionado.csv'), index=False)
print(f"\nGuardado. Filas del modelo: {filas_modelo} | filas totales: {len(df)}")
```

```python
#Celda 14 -> Normalización
import pandas as pd
import numpy as np
import os

df_part = pd.read_csv(os.path.join(ruta_salida, 'dataset_1hz_particionado.csv'), parse_dates=['timestamp'])

VARIABLES_MODELO = [
    'rpm', 'carga_motor_pct', 'temp_refrigerante_C', 'temp_aire_admision_C',
    'presion_multiple_kPa', 'acelerador_b_pct', 'afr', 'velocidad_obd_kmh',
    'voltaje_adaptador_V', 'velocidad_gps_kmh', 'acel_total_ms2'
]
PARTICIONES_MODELO = ['entrenamiento', 'validacion', 'prueba']

# Comprobación previa: las variables existen y no hay vacíos en las filas del modelo
assert set(VARIABLES_MODELO) <= set(df_part.columns), "Falta alguna variable en el dataset."
modelo = df_part['particion'].isin(PARTICIONES_MODELO)
assert df_part.loc[modelo, VARIABLES_MODELO].notna().all().all(), "Hay vacíos en las filas del modelo."

# Media y desviación solo con entrenamiento
entrenamiento = df_part[df_part['particion'] == 'entrenamiento']
media = entrenamiento[VARIABLES_MODELO].mean()
desviacion = entrenamiento[VARIABLES_MODELO].std()
assert (desviacion > 0).all(), "Hay variables con desviación cero en entrenamiento."

# Aplicar la misma escala a todas las filas
df_norm = df_part.copy()
df_norm[VARIABLES_MODELO] = (df_norm[VARIABLES_MODELO] - media) / desviacion

# Comprobaciones: entrenamiento con media 0 y desviación 1; sin infinitos
z_train = df_norm.loc[df_norm['particion'] == 'entrenamiento', VARIABLES_MODELO]
assert np.allclose(z_train.mean(), 0, atol=1e-6), "La media en entrenamiento no es 0."
assert np.allclose(z_train.std(), 1, atol=1e-6), "La desviación en entrenamiento no es 1."
assert not np.isinf(df_norm[VARIABLES_MODELO].to_numpy()).any(), "Hay valores infinitos."

# Guardar parámetros y el dataset
parametros = pd.DataFrame({'variable': VARIABLES_MODELO, 'media': media.values, 'desviacion': desviacion.values})
parametros.to_csv(os.path.join(ruta_salida, 'parametros_normalizacion.csv'), index=False)
df_norm.to_csv(os.path.join(ruta_salida, 'dataset_normalizado.csv'), index=False)

display(parametros.round(2))
print("\nMedia de las variables normalizadas por partición:")
display(df_norm.groupby('particion')[VARIABLES_MODELO].mean().round(2).T)
print(f"\nGuardados: parametros_normalizacion.csv y dataset_normalizado.csv ({len(df_norm)} filas)")
```

```python
# Celda 15 -> Generación de características móviles y ventanas 3D
import pandas as pd
import os
import numpy as np

# Parámetros
VENTANA_SEG = 30
HORIZONTE_SEG = 3
VENTANA_MOVIL_SEG = 10

# Listas de variables
VARIABLES_MODELO = [
    'rpm', 'carga_motor_pct', 'temp_refrigerante_C', 'temp_aire_admision_C',
    'presion_multiple_kPa', 'acelerador_b_pct', 'afr', 'velocidad_obd_kmh',
    'voltaje_adaptador_V', 'velocidad_gps_kmh', 'acel_total_ms2'
]

VARIABLES_MOVILES = ['rpm', 'carga_motor_pct', 'acelerador_b_pct', 'velocidad_gps_kmh']
PARTICIONES_MODELO = ['entrenamiento', 'validacion', 'prueba']

# Cargar el dataset normalizado
df_norm = pd.read_csv(os.path.join(ruta_salida, 'dataset_normalizado.csv'), parse_dates=['timestamp'])
df_modelo = df_norm[df_norm['particion'].isin(PARTICIONES_MODELO)].copy()
df_modelo = df_modelo.sort_values(['viaje_num', 'segmento_id', 'timestamp']).reset_index(drop=True)

nuevas_columnas = []

# Características móviles, por segmento
for var in VARIABLES_MOVILES:
    # Media móvil de 10 segundos
    col_mean = f'{var}_mean10'
    df_modelo[col_mean] = df_modelo.groupby(['viaje_num', 'segmento_id'])[var].transform(lambda x: x.rolling(VENTANA_MOVIL_SEG).mean())

    # Desviación móvil de 10 segundos
    col_std = f'{var}_std10'
    df_modelo[col_std] = df_modelo.groupby(['viaje_num', 'segmento_id'])[var].transform(lambda x: x.rolling(VENTANA_MOVIL_SEG).std())

    # Diferencia de 1 segundo (por segmento)
    col_diff = f'{var}_diff1'
    df_modelo[col_diff] = df_modelo.groupby(['viaje_num', 'segmento_id'])[var].transform(lambda x: x.diff(1))

    # Llenar la lista
    nuevas_columnas.extend([col_mean, col_std, col_diff])

# Definir la lista de las 23 características
CARACTERISTICAS = VARIABLES_MODELO + nuevas_columnas

# Eliminar los vacíos generados por las ventanas móviles
df_modelo = df_modelo.dropna(subset=CARACTERISTICAS).reset_index(drop=True)

x_list = {'entrenamiento': [], 'validacion': [], 'prueba': []}
y_list = {'entrenamiento': [], 'validacion': [], 'prueba': []}

# Agrupar por segmento
for (viaje, segmento), df_seg in df_modelo.groupby(['viaje_num', 'segmento_id']):
    particion = df_seg['particion'].iloc[0]

    # Extraer matriz de 23 características y vector objetivo (carga_motor_pct)
    data_x = df_seg[CARACTERISTICAS].values
    data_y = df_seg['carga_motor_pct'].values

    n_filas = len(df_seg)

    # Generar ventanas
    for i in range(n_filas - VENTANA_SEG - HORIZONTE_SEG + 1):
        x_win = data_x[i : i + VENTANA_SEG]
        objetivo = data_y[i + VENTANA_SEG + HORIZONTE_SEG - 1]

        x_list[particion].append(x_win)
        y_list[particion].append(objetivo)

# Convertir a arreglos NumPy
X_train = np.array(x_list['entrenamiento'])
y_train = np.array(y_list['entrenamiento'])

X_val = np.array(x_list['validacion'])
y_val = np.array(y_list['validacion'])

X_test = np.array(x_list['prueba'])
y_test = np.array(y_list['prueba'])

# Validar que no existan valores vacíos (NaN) en los arreglos
for nombre, arreglo in [('X_train', X_train), ('y_train', y_train),
                        ('X_val', X_val), ('y_val', y_val),
                        ('X_test', X_test), ('y_test', y_test)]:
    assert not np.isnan(arreglo).any(), f"Hay valores vacíos (NaN) en {nombre}"

# Forma: se ajusta a los parámetros
for nombre, X in [('X_train', X_train), ('X_val', X_val), ('X_test', X_test)]:
    assert X.shape[1:] == (VENTANA_SEG, len(CARACTERISTICAS)), f"Forma incorrecta en {nombre}: {X.shape}"

# Conteo de ventanas
print(f"Entrenamiento: {len(X_train)} ventanas")
print(f"Validación:    {len(X_val)} ventanas")
print(f"Prueba:        {len(X_test)} ventanas")
print(f"Total:         {len(X_train) + len(X_val) + len(X_test)} ventanas")
print(f"Forma de X:    {X_train.shape[1:]} (ventana, características)")

# Conteo de referencia: solo se exige con la configuración de referencia
if (VENTANA_SEG, HORIZONTE_SEG, VENTANA_MOVIL_SEG) == (30, 3, 10):
    assert len(X_train) == 6959, f"Esperados 6959 en train, obtenidos {len(X_train)}"
    assert len(X_val) == 1343, f"Esperados 1343 en val, obtenidos {len(X_val)}"
    assert len(X_test) == 2342, f"Esperados 2342 en test, obtenidos {len(X_test)}"

# Guardar en archivo comprimido .npz
ruta_npz = os.path.join(ruta_salida, 'dataset_preparado_3d.npz')

np.savez_compressed(
    ruta_npz,
    X_train=X_train, y_train=y_train,
    X_val=X_val, y_val=y_val,
    X_test=X_test, y_test=y_test,
    columnas=np.array(CARACTERISTICAS),
    parametros=np.array([VENTANA_SEG, HORIZONTE_SEG, VENTANA_MOVIL_SEG])
)

print(f"\nDataset guardado en '{ruta_npz}'")
```

```python
# Celda 16 -> Verificación final y diccionario de datos (E3)
import os
import numpy as np
import pandas as pd

# Rutas de entrada
ruta_salida = '/content/drive/MyDrive/Colab Notebooks/dataset/dataset_limpio'
ruta_npz = os.path.join(ruta_salida, 'dataset_preparado_3d.npz')
ruta_params = os.path.join(ruta_salida, 'parametros_normalizacion.csv')
ruta_norm = os.path.join(ruta_salida, 'dataset_normalizado.csv')

# Cargar archivos de forma independiente
datos_3d = np.load(ruta_npz)
X_train_loaded = datos_3d['X_train']
y_train_loaded = datos_3d['y_train']
X_val_loaded = datos_3d['X_val']
y_val_loaded = datos_3d['y_val']
X_test_loaded = datos_3d['X_test']
y_test_loaded = datos_3d['y_test']
columnas_loaded = datos_3d['columnas']
parametros_loaded = datos_3d['parametros']

params_df = pd.read_csv(ruta_params)
df_norm_loaded = pd.read_csv(ruta_norm, parse_dates=['timestamp'])

# 1. Comprobación de Formas y Vacíos
assert X_train_loaded.shape == (6959, 30, 23), f"Forma X_train incorrecta: {X_train_loaded.shape}"
assert X_val_loaded.shape == (1343, 30, 23), f"Forma X_val incorrecta: {X_val_loaded.shape}"
assert X_test_loaded.shape == (2342, 30, 23), f"Forma X_test incorrecta: {X_test_loaded.shape}"
assert y_train_loaded.shape == (6959,), f"Forma y_train incorrecta: {y_train_loaded.shape}"
assert y_val_loaded.shape == (1343,), f"Forma y_val incorrecta: {y_val_loaded.shape}"
assert y_test_loaded.shape == (2342,), f"Forma y_test incorrecta: {y_test_loaded.shape}"

assert not np.isnan(X_train_loaded).any(), "Vacíos detectados en X_train"
assert not np.isnan(X_val_loaded).any(), "Vacíos detectados en X_val"
assert not np.isnan(X_test_loaded).any(), "Vacíos detectados en X_test"
assert not np.isnan(y_train_loaded).any(), "Vacíos detectados en y_train"
assert not np.isnan(y_val_loaded).any(), "Vacíos detectados en y_val"
assert not np.isnan(y_test_loaded).any(), "Vacíos detectados en y_test"

assert len(columnas_loaded) == 23, f"Esperados 23 nombres de columnas, obtenidos {len(columnas_loaded)}"
assert np.array_equal(parametros_loaded, [30, 3, 10]), f"Parámetros de ventana incorrectos: {parametros_loaded}"

# 2. Comprobación de Particiones Disjuntas

df_modelo = df_norm_loaded[df_norm_loaded['particion'].isin(['entrenamiento', 'validacion', 'prueba'])]
assert df_modelo.groupby('viaje_num')['particion'].nunique().max() == 1, "Hay viajes asignados a más de una partición"

part_viajes = df_modelo.groupby('particion')['viaje_num'].unique().to_dict()
assert set(part_viajes['entrenamiento']) == {1, 2, 3, 4, 8, 9}, "Fallo en viajes de entrenamiento"
assert set(part_viajes['validacion']) == {7}, "Fallo en viajes de validación"
assert set(part_viajes['prueba']) == {5, 6}, "Fallo en viajes de prueba"

# 3. Comprobación de Continuidad en los Segmentos
df_rec_clean = df_modelo.sort_values(['viaje_num', 'segmento_id', 'timestamp']).copy()
dt_seg = df_rec_clean.groupby(['viaje_num', 'segmento_id'])['timestamp'].diff().dt.total_seconds().dropna()
assert (dt_seg == 1.0).all(), f"Existen saltos de tiempo distintos a 1.0 s en los segmentos reorganizados: {dt_seg.unique()}"

# 4. Comprobación de Ventana Revisada a Mano
primer_seg = df_rec_clean[(df_rec_clean['viaje_num'] == 1) & (df_rec_clean['segmento_id'] == 1)].sort_values('timestamp')
sub_X_esperado = primer_seg.iloc[9:39][columnas_loaded[:11]].values
sub_X_ventana = X_train_loaded[0][:, :11]
assert np.allclose(sub_X_ventana, sub_X_esperado, atol=1e-5), "La matriz X de la primera ventana no coincide con las filas de origen"

sub_y_esperado = primer_seg.iloc[41]['carga_motor_pct']
sub_y_ventana = y_train_loaded[0]
assert np.isclose(sub_y_ventana, sub_y_esperado, atol=1e-5), "El valor objetivo y no coincide con el horizonte +3 esperado"


# 5. Objetivo en Escala Real
media_carga = params_df.loc[params_df['variable'].str.strip() == 'carga_motor_pct', 'media'].values[0]
desv_carga = params_df.loc[params_df['variable'].str.strip() == 'carga_motor_pct', 'desviacion'].values[0]

y_train_real = y_train_loaded * desv_carga + media_carga
y_val_real = y_val_loaded * desv_carga + media_carga
y_test_real = y_test_loaded * desv_carga + media_carga

assert (y_train_real >= -1e-6).all() and (y_train_real <= 100 + 1e-6).all(), f"Fallo rango real train: {y_train_real.min()} a {y_train_real.max()}"
assert (y_val_real >= -1e-6).all() and (y_val_real <= 100 + 1e-6).all(), f"Fallo rango real val: {y_val_real.min()} a {y_val_real.max()}"
assert (y_test_real >= -1e-6).all() and (y_test_real <= 100 + 1e-6).all(), f"Fallo rango real test: {y_test_real.min()} a {y_test_real.max()}"

assert np.isclose(y_train_real.mean(), 22.1, atol=0.5), f"Media real train incorrecta: {y_train_real.mean():.2f}"
assert np.isclose(y_val_real.mean(), 23.9, atol=0.5), f"Media real val incorrecta: {y_val_real.mean():.2f}"
assert np.isclose(y_test_real.mean(), 23.7, atol=0.5), f"Media real test incorrecta: {y_test_real.mean():.2f}"

print("Verificación completa")

# Generación del Diccionario de Datos
plantilla_tratamiento = "Vacíos (-) a NaN; promedio por segundo (de 2 Hz a 1 Hz); interpolación lineal hasta 3 s dentro del segmento; z-score con entrenamiento"

variables_base_info = {
    'rpm': {
        'descripcion': "Revoluciones por minuto del motor",
        'columna_original': "Engine RPM(rpm)",
        'unidad_original': "rpm",
        'unidad_fisica': "rpm",
        'unidad_final': "z",
        'tipo': "motor",
        'tratamiento': plantilla_tratamiento
    },
    'carga_motor_pct': {
        'descripcion': "Porcentaje de la carga del motor",
        'columna_original': "Engine Load(%)",
        'unidad_original': "%",
        'unidad_fisica': "%",
        'unidad_final': "z",
        'tipo': "motor",
        'tratamiento': plantilla_tratamiento
    },
    'temp_refrigerante_C': {
        'descripcion': "Temperatura del refrigerante del motor",
        'columna_original': "Engine Coolant Temperature(°F)",
        'unidad_original': "°F",
        'unidad_fisica': "°C",
        'unidad_final': "z",
        'tipo': "motor",
        'tratamiento': "Conversión de °F a °C; " + plantilla_tratamiento
    },
    'temp_aire_admision_C': {
        'descripcion': "Temperatura del aire de admisión",
        'columna_original': "Intake Air Temperature(°F)",
        'unidad_original': "°F",
        'unidad_fisica': "°C",
        'unidad_final': "z",
        'tipo': "motor",
        'tratamiento': "Conversión de °F a °C; " + plantilla_tratamiento
    },
    'presion_multiple_kPa': {
        'descripcion': "Presión absoluta en el múltiple de admisión",
        'columna_original': "Intake Manifold Pressure(psi)",
        'unidad_original': "psi",
        'unidad_fisica': "kPa",
        'unidad_final': "z",
        'tipo': "motor",
        'tratamiento': "Conversión de psi a kPa; " + plantilla_tratamiento
    },
    'acelerador_b_pct': {
        'descripcion': "Posición absoluta del acelerador en el canal B",
        'columna_original': "Absolute Throttle Position B(%)",
        'unidad_original': "%",
        'unidad_fisica': "%",
        'unidad_final': "z",
        'tipo': "motor",
        'tratamiento': plantilla_tratamiento
    },
    'afr': {
        'descripcion': "Relación aire-combustible medida",
        'columna_original': "Air Fuel Ratio(Measured)(:1)",
        'unidad_original': ":1",
        'unidad_fisica': ":1",
        'unidad_final': "z",
        'tipo': "motor",
        'tratamiento': plantilla_tratamiento
    },
    'velocidad_obd_kmh': {
        'descripcion': "Velocidad registrada por el puerto OBD",
        'columna_original': "Speed (OBD)(mph)",
        'unidad_original': "mph",
        'unidad_fisica': "km/h",
        'unidad_final': "z",
        'tipo': "velocidad",
        'tratamiento': "Conversión de mph a km/h; " + plantilla_tratamiento
    },
    'voltaje_adaptador_V': {
        'descripcion': "Voltaje eléctrico detectado por el adaptador OBD",
        'columna_original': "Voltage (OBD Adapter)(V)",
        'unidad_original': "V",
        'unidad_fisica': "V",
        'unidad_final': "z",
        'tipo': "motor",
        'tratamiento': plantilla_tratamiento
    },
    'velocidad_gps_kmh': {
        'descripcion': "Velocidad derivada del sensor GPS del dispositivo móvil",
        'columna_original': "GPS Speed (Meters/second)",
        'unidad_original': "m/s",
        'unidad_fisica': "km/h",
        'unidad_final': "z",
        'tipo': "velocidad",
        'tratamiento': "Conversión de m/s a km/h; " + plantilla_tratamiento
    },
    'acel_total_ms2': {
        'descripcion': "Aceleración total del dispositivo sin la fuerza de gravedad",
        'columna_original': "G(x), G(y), G(z)",
        'unidad_original': "m/s²",
        'unidad_fisica': "m/s²",
        'unidad_final': "z",
        'tipo': "aceleración",
        'tratamiento': "Fórmula √(gx² + gy² + gz²) − 9.80665 sobre sensores brutos; " + plantilla_tratamiento
    }
}

variables_derivadas_lista = []
for var in ['rpm', 'carga_motor_pct', 'acelerador_b_pct', 'velocidad_gps_kmh']:
    variables_derivadas_lista.append({
        'variable': f"{var}_mean10",
        'descripcion': f"Media de los últimos 10 s de {var}, dentro del segmento",
        'columna_original': "",
        'unidad_original': "",
        'unidad_fisica': "z",
        'unidad_final': "z",
        'tipo': "derivada",
        'viajes_con_dato_de_18': np.nan,
        'pct_nulos': np.nan,
        'tratamiento': f"Media móvil calculada sobre ventana de 10 muestras de {var} normalizada, por segmento",
        'media': "",
        'desviacion': ""
    })
    variables_derivadas_lista.append({
        'variable': f"{var}_std10",
        'descripcion': f"Desviación estándar de los últimos 10 s de {var}, dentro del segmento",
        'columna_original': "",
        'unidad_original': "",
        'unidad_fisica': "z",
        'unidad_final': "z",
        'tipo': "derivada",
        'viajes_con_dato_de_18': np.nan,
        'pct_nulos': np.nan,
        'tratamiento': f"Desviación móvil calculada sobre ventana de 10 muestras de {var} normalizada, por segmento",
        'media': "",
        'desviacion': ""
    })
    variables_derivadas_lista.append({
        'variable': f"{var}_diff1",
        'descripcion': f"Cambio instantáneo (1 s) de la variable {var}, dentro del segmento",
        'columna_original': "",
        'unidad_original': "",
        'unidad_fisica': "z",
        'unidad_final': "z",
        'tipo': "derivada",
        'viajes_con_dato_de_18': np.nan,
        'pct_nulos': np.nan,
        'tratamiento': f"Diferencia de primer orden calculada sobre la variable {var} normalizada, por segmento",
        'media': "",
        'desviacion': ""
    })

df_derivadas = pd.DataFrame(variables_derivadas_lista)

perfil_var_df = pd.read_csv(os.path.join(ruta_salida, 'perfil_variables.csv'))

base_filas = []
for var, info in variables_base_info.items():
    perfil_row = perfil_var_df[perfil_var_df['variable'] == var]
    viajes_con_dato = perfil_row['viajes_con_dato'].values[0] if len(perfil_row) and 'viajes_con_dato' in perfil_row.columns else (perfil_row['viajes_con_dato_de_18'].values[0] if len(perfil_row) and 'viajes_con_dato_de_18' in perfil_row.columns else 0)
    pct_nulos = perfil_row['pct_nulos'].values[0] if len(perfil_row) else 0.0

    param_row = params_df[params_df['variable'] == var]
    media = round(param_row['media'].values[0], 4) if len(param_row) else ""
    desviacion = round(param_row['desviacion'].values[0], 4) if len(param_row) else ""

    base_filas.append({
        'variable': var,
        'descripcion': info['descripcion'],
        'columna_original': info['columna_original'],
        'unidad_original': info['unidad_original'],
        'unidad_fisica': info['unidad_fisica'],
        'unidad_final': info['unidad_final'],
        'tipo': info['tipo'],
        'viajes_con_dato_de_18': viajes_con_dato,
        'pct_nulos': pct_nulos,
        'tratamiento': info['tratamiento'],
        'media': media,
        'desviacion': desviacion
    })
df_base_completas = pd.DataFrame(base_filas)

for i, row in df_derivadas.iterrows():
    base_name = row['variable'].split('_mean10')[0].split('_std10')[0].split('_diff1')[0]
    correspondiente = df_base_completas[df_base_completas['variable'] == base_name]
    if len(correspondiente):
        df_derivadas.at[i, 'viajes_con_dato_de_18'] = correspondiente['viajes_con_dato_de_18'].values[0]
        df_derivadas.at[i, 'pct_nulos'] = correspondiente['pct_nulos'].values[0]

diccionario_datos = pd.concat([df_base_completas, df_derivadas], ignore_index=True)
diccionario_datos = diccionario_datos.set_index('variable').reindex(columnas_loaded).reset_index()

assert len(diccionario_datos) == 23, f"El diccionario debe tener exactamente 23 filas, tiene: {len(diccionario_datos)}"
assert diccionario_datos['variable'].nunique() == 23, "Existen nombres duplicados en el diccionario de datos"

for col in ['descripcion', 'unidad_fisica', 'unidad_final', 'tipo', 'viajes_con_dato_de_18', 'pct_nulos', 'tratamiento']:
    assert diccionario_datos[col].notna().all() and (diccionario_datos[col] != "").all(), f"Campo '{col}' vacío detectado"

diccionario_datos.to_csv(os.path.join(ruta_salida, 'diccionario_datos.csv'), index=False, encoding='utf-8-sig')
print("Diccionario de datos guardado correctamente con 23 variables.")


# Archivo de Decisiones y Supuestos
decisiones_datos = [
    {"parámetro": "Recorte inicial por viaje", "valor": "10 s", "motivo": "Precaución: en el viaje 1 las primeras filas mostraron valores iniciales atípicos; no se comprobó la causa física"},
    {"parámetro": "Umbral de huecos", "valor": "5 s", "motivo": "Un salto de tiempo mayor a este umbral abre automáticamente un segmento nuevo"},
    {"parámetro": "Frecuencia común", "valor": "1 Hz", "motivo": "Permite unificar viajes originalmente muestreados a diferentes frecuencias de 2 Hz y 1 Hz"},
    {"parámetro": "Interpolación", "valor": "Hasta 3 s", "motivo": "Los vacíos cortos son tratados como ruido y se rellenan mediante interpolación lineal dentro de cada segmento"},
    {"parámetro": "Segmento mínimo", "valor": "30 s", "motivo": "Longitud mínima indispensable requerida para poder extraer al menos una ventana completa de análisis"},
    {"parámetro": "Partición", "valor": "Entrenamiento 1, 2, 3, 4, 8, 9; validación 7; prueba 5, 6", "motivo": "División por viajes independientes, manteniendo juntos aquellos viajes realizados en el mismo día para evitar fuga de datos"},
    {"parámetro": "Normalización", "valor": "z-score con entrenamiento", "motivo": "Evitar fugar información de prueba calculando las estadísticas de normalización únicamente con entrenamiento"},
    {"parámetro": "Ventana, horizonte y móviles", "valor": "30 s, 3 s, 10 s", "motivo": "Elegida con la autocorrelación de la carga y una prueba rápida con modelos simples; es ajustable"},
    {"parámetro": "Variables", "valor": "11", "motivo": "Uso de acel_total_ms2 unificada para evitar el efecto de rotación de G(x/y/z) y acelerador_b_pct como respaldo de acelerador principal"},
    {"parámetro": "Objetivo", "valor": "carga_motor_pct a +3 s", "motivo": "Optimizar el horizonte temporal para lograr un pronóstico útil y con antelación de la carga física real del motor"},
    {"parámetro": "Limitaciones", "valor": "Sin etiquetas de falla, volumen limitado y desequilibrio", "motivo": "Falta de etiquetas explícitas de falla. Solo se dispone de 9 viajes completos con motor (~3 h totales). El viaje 16 es casi puro ralentí y la variable temp_aire_admision_C está sesgada por la temperatura ambiental del día."}
]

df_decisiones = pd.DataFrame(decisiones_datos)
df_decisiones.to_csv(os.path.join(ruta_salida, 'decisiones_supuestos.csv'), index=False, encoding='utf-8-sig')
print("Decisiones y supuestos guardados correctamente.")
```
